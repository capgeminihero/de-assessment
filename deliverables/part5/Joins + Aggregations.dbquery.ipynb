```sql
-- Episodes per season with show name joined
SELECT
    s.show_name,
    e.season,
    COUNT(e.episode_id)                    AS episode_count,
    ROUND(AVG(e.runtime), 1)               AS avg_runtime_mins,
    RANK() OVER (ORDER BY COUNT(e.episode_id) DESC) AS season_size_rank
FROM de_assessment_dev.silver.silver_episodes e
JOIN de_assessment_dev.silver.silver_shows s ON e.show_id = s.show_id
WHERE e.runtime > 0
GROUP BY s.show_name, e.season
ORDER BY episode_count DESC;
```
