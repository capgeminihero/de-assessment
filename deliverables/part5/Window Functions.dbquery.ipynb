```sql
-- Rank episodes within each season by runtime
SELECT
    show_id,
    season,
    episode_id,
    episode_name,
    runtime,
    ROW_NUMBER() OVER (PARTITION BY show_id, season ORDER BY runtime DESC) AS runtime_rank,
    AVG(runtime) OVER (PARTITION BY show_id, season) AS avg_season_runtime
FROM de_assessment_dev.silver.silver_episodes
WHERE runtime IS NOT NULL
  AND season IS NOT NULL 
ORDER BY show_id, season, runtime_rank;
```
